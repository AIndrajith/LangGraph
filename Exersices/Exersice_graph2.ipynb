### Exercise for Graph II

Create a **Graph** where you pass in a single list of integers along with a name and an operation. If the operation is a "+", you **add** the elements and if it is a "*", you **multiply** the elements, **all within the same node**.

- **Input**: {"name":"Jack Sparrow", "values":[1,2,3,4], "operation":"*"}
- **Output**:"Hi Jack Sparrow, your answer is : 24"

__Hint__: You need an if-statement in your node!

```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values : List[int]
    name : str
    operation : str
    result: str
```

```python
def total(operation: str, values : List[int]) -> int:
    """this one handles the operation in values"""
    global tot
    if operation == "+":
        tot = 0
        for i in values:
            tot += i
    elif operation == "*":
        tot = 1
        for i in values:
            tot *= i
    else:
        print("Invalid operation!")

    return tot
```

```python
def process_values(state:AgentState) -> AgentState:
    """This one handles multiple different Inputs"""

    state["result"] = f"Hi {state["name"]} Your answer is {total(state["operation"], state["values"])}"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("processor", process_values)
graph.set_entry_point("processor")
graph.set_finish_point("processor")

app = graph.compile()
```

```python
answer = app.invoke({"name":"Jack Sparrow", "values":[1,2,3,4], "operation":"+"})
```

```python
print(answer["result"])
```
