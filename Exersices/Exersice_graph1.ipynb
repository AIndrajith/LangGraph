### Exercise 1

Create a **Personalized Compliment Agent** using LangGraph!

- **Input**:{"name":"Bob"}
- **Output**:"Bob, you're doing an amazing job Learning LangGraph!"

__Hint__: You have to concatenate the state, not replace it!

```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name : str
    message : str

def greeting_node(state: AgentState) -> AgentState:
    state['message'] = state['name'] + ", you're doing and amazing job learning LangGraph!"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("greeter", greeting_node)

graph.set_entry_point("greeter")
graph.set_finish_point("greeter")

app = graph.compile()
```

```python
result = app.invoke({"name":"Bob"})
```
